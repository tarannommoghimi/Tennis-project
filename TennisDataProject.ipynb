```python
import pandas as pd
BASE = r"C:\Users\Personal Camputer\OneDrive\Desktop\tennis"

pbp_df        = pd.read_csv(BASE + r"\pbp.csv")

# ==========================================
# قدم ۲: حذف تکراری‌ها (همون الگوی همیشگی، رو هر فایل جدا)
# ==========================================
home_score_df = home_score.drop_duplicates()
home_score_df = home_score.drop_duplicates(subset='match_id', keep='first')

away_score_df = away_score.drop_duplicates()
away_score_df = away_score.drop_duplicates(subset='match_id', keep='first')

home_team_df = home_team.drop_duplicates()
home_team_df = home_team.drop_duplicates(subset='match_id', keep='first')

away_team_df = away_team.drop_duplicates()
away_team_df = away_team.drop_duplicates(subset='match_id', keep='first')

# ==========================================
# قدم ۳: تبدیل period_1..5 به فرمت long (هر ست یه ردیف) با melt
# ==========================================
home_sets = home_score_df.melt(
    id_vars='match_id',
    value_vars=['period_1', 'period_2', 'period_3', 'period_4', 'period_5'],
    var_name='period_col',
    value_name='home_games'
)
home_sets['set_id'] = home_sets['period_col'].str.extract(r'(\d)')
home_sets['set_id'] = home_sets['set_id'].astype(int)
home_sets = home_sets.drop(columns='period_col')
home_sets = home_sets.dropna(subset=['home_games'])
home_sets['home_games'] = pd.to_numeric(home_sets['home_games'], errors='coerce')
home_sets = home_sets.dropna(subset=['home_games'])
home_sets['home_games'] = home_sets['home_games'].astype(int)

away_sets = away_score_df.melt(
    id_vars='match_id',
    value_vars=['period_1', 'period_2', 'period_3', 'period_4', 'period_5'],
    var_name='period_col',
    value_name='away_games'
)
away_sets['set_id'] = away_sets['period_col'].str.extract(r'(\d)')
away_sets['set_id'] = away_sets['set_id'].astype(int)
away_sets = away_sets.drop(columns='period_col')
away_sets = away_sets.dropna(subset=['away_games'])
away_sets['away_games'] = pd.to_numeric(away_sets['away_games'], errors='coerce')
away_sets = away_sets.dropna(subset=['away_games'])
away_sets['away_games'] = away_sets['away_games'].astype(int)

# ترکیب گیم‌های home و away برای هر ست
set_scores = home_sets.merge(away_sets, on=['match_id', 'set_id'])
set_scores['total_games'] = set_scores['home_games'] + set_scores['away_games']

# ==========================================
# قدم ۴: اعتبارسنجی نمره‌ی ست
# فقط نمره‌هایی که واقعاً در تنیس ممکنن رو نگه می‌داریم:
#   6-0 تا 6-4   (ست عادی)
#   7-5          (ست با یه گیم اضافه)
#   7-6          (ست با تای‌بریک)
#   1-0          (match tie-break، فقط اگه ست آخرِ match باشه)
# ==========================================
h = set_scores['home_games']
a = set_scores['away_games']

standard_set = ((h == 6) & (a >= 0) & (a <= 4)) | ((a == 6) & (h >= 0) & (h <= 4))
set_7_5 = ((h == 7) & (a == 5)) | ((a == 7) & (h == 5))
set_7_6 = ((h == 7) & (a == 6)) | ((a == 7) & (h == 6))

# مشخص کردن ست آخر هر match
last_set_per_match = set_scores.groupby('match_id')['set_id'].max()
last_set_per_match = last_set_per_match.reset_index()
last_set_per_match = last_set_per_match.rename(columns={'set_id': 'last_set_id'})

set_scores = set_scores.merge(last_set_per_match, on='match_id')
is_last_set = set_scores['set_id'] == set_scores['last_set_id']

match_tiebreak = is_last_set & (((h == 1) & (a == 0)) | ((a == 1) & (h == 0)))

valid = standard_set | set_7_5 | set_7_6 | match_tiebreak
set_scores['is_valid'] = valid
set_scores['is_match_tiebreak'] = match_tiebreak

print("ست‌ها قبل از اعتبارسنجی:", len(set_scores))
set_scores_valid = set_scores[set_scores['is_valid']].copy()
print("ست‌ها بعد از اعتبارسنجی:", len(set_scores_valid))
print("ست‌های حذف‌شده (نمره‌ی غیرممکن):", len(set_scores) - len(set_scores_valid))

# ==========================================
# قدم ۵: اعتبارسنجی متقابل با pbp.csv
# تعداد گیم واقعی هر ست رو از point-by-point هم می‌شماریم
# فقط ستی که هم تو score هم تو pbp باشه قبول میشه
# ==========================================
pbp_clean = pbp_df.dropna(subset=['match_id', 'set_id', 'game_id']).copy()
pbp_clean['match_id'] = pbp_clean['match_id'].astype(int)

games_from_pbp = pbp_clean.groupby(['match_id', 'set_id'])['game_id'].nunique()
games_from_pbp = games_from_pbp.reset_index()
games_from_pbp = games_from_pbp.rename(columns={'game_id': 'num_games_pbp'})

final_sets = set_scores_valid.merge(games_from_pbp, on=['match_id', 'set_id'], how='inner')

# ==========================================
# قدم ۶: اضافه کردن جنسیت (فقط matchهایی که هر دو بازیکن هم‌جنسن)
# ==========================================
home_gender = home_team_df[['match_id', 'gender']].dropna().copy()
home_gender['gender'] = home_gender['gender'].str.strip().str.upper()
home_gender = home_gender.rename(columns={'gender': 'home_gender'})

away_gender = away_team_df[['match_id', 'gender']].dropna().copy()
away_gender['gender'] = away_gender['gender'].str.strip().str.upper()
away_gender = away_gender.rename(columns={'gender': 'away_gender'})

gender_df = home_gender.merge(away_gender, on='match_id')
gender_df = gender_df[gender_df['home_gender'] == gender_df['away_gender']]
gender_df = gender_df[gender_df['home_gender'].isin(['M', 'F'])]
gender_df['gender'] = gender_df['home_gender']
gender_df = gender_df[['match_id', 'gender']]

analysis_df = final_sets.merge(gender_df, on='match_id')

print("\nتعداد کل رکورد (match, set) نهایی:", len(analysis_df))
print("تعداد match یکتا:", analysis_df['match_id'].nunique())

# ==========================================
# قدم ۷: میانگین گیم به‌ازای هر ست، به تفکیک جنسیت
# match tie-break (ست‌های ۱ گیمی) کنار گذاشته میشه چون ساختارش فرق داره
# ==========================================
normal_sets = analysis_df[analysis_df['is_match_tiebreak'] == False]

result = normal_sets.groupby('gender')['total_games'].agg(['mean', 'median', 'std', 'count'])
result = result.round(3)
result = result.rename(index={'M': 'Men', 'F': 'Women'})

print("\n=== میانگین تعداد گیم در هر ست: مردان در مقابل زنان ===")
print(result)
```
