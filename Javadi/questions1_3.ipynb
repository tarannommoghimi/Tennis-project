# Question 1

```python
df_home = pd.read_csv("../home_team.csv")
df_away = pd.read_csv("../away_team.csv")
```

```python
all_players = pd.concat([
    df_home[['player_id', 'full_name' , 'height']],
    df_away[['player_id', 'full_name' , 'height']]
], ignore_index=True)
all_players
```

```python
all_players = all_players.dropna(subset=['player_id'])
unique_players = all_players.drop_duplicates(subset='player_id')
unique_players.sort_values(["player_id"])
```

```python
print(f"Total number of players: {unique_players["player_id"].count()}")
```

```python
average_height = unique_players["height"].mean().round(2)
print(f"The average height of players is: {average_height}")
```

```python
df = pd.read_csv("../event.csv")
df.dropna(subset=["winner_code"])
```

```python
df_event = pd.read_csv("../event.csv")
```

```python
player_id = 14882 #Novak Djokovic 

home_matches = df_home[df_home['player_id'] == player_id]['match_id']
away_matches = df_away[df_away['player_id'] == player_id]['match_id']

all_match_ids = pd.concat([home_matches, away_matches]).unique()

player_matches = df_event[df_event['match_id'].isin(all_match_ids)]

player_matches = player_matches.merge(df_home[['match_id', 'player_id', 'full_name']], on='match_id', how='left') \
                               .rename(columns={'player_id': 'home_player_id', 'full_name': 'home_full_name'})

player_matches = player_matches.merge(df_away[['match_id', 'player_id', 'full_name']], on='match_id', how='left') \
                               .rename(columns={'player_id': 'away_player_id', 'full_name': 'away_full_name'})
```

```python
player_matches
```

```python
df_event_clean = df_event[['match_id', 'winner_code' , 'start_datetime']].drop_duplicates(subset=['match_id'])
df_home_clean = df_home[['match_id', 'player_id', 'full_name']].drop_duplicates(subset=['match_id'])
df_away_clean = df_away[['match_id', 'player_id', 'full_name']].drop_duplicates(subset=['match_id'])
```

```python
df_winner = (df_event_clean[['match_id', 'winner_code']]
    .merge(df_home_clean[['match_id', 'player_id', 'full_name']], on='match_id', how='left')
    .rename(columns={'player_id': 'home_player_id', 'full_name': 'home_full_name'})
    .merge(df_away_clean[['match_id', 'player_id', 'full_name']], on='match_id', how='left')
    .rename(columns={'player_id': 'away_player_id', 'full_name': 'away_full_name'}))
df_winner
```

```python
home_wins = (df_winner[df_winner['winner_code'] == 1]
             .dropna(subset=['home_player_id', 'home_full_name'])
             .groupby(['home_player_id', 'home_full_name']).size().reset_index(name='wins'))
home_wins = home_wins.rename(columns={'home_player_id': 'player_id', 'home_full_name': 'full_name'})

away_wins = (df_winner[df_winner['winner_code'] == 2]
             .dropna(subset=['away_player_id', 'away_full_name'])
             .groupby(['away_player_id', 'away_full_name']).size().reset_index(name='wins'))
away_wins = away_wins.rename(columns={'away_player_id': 'player_id', 'away_full_name': 'full_name'})

away_wins
```

```python
total_wins = (pd.concat([home_wins, away_wins])
               .groupby(['player_id', 'full_name'])['wins']
               .sum()
               .reset_index()
               .sort_values('wins', ascending=False))

print(total_wins.head(1))
```
