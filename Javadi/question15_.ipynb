```python
import pandas as pd
```

```python
home_df  = pd.read_csv("../home_team.csv")
away_df  = pd.read_csv("../away_team.csv")
venue_df = pd.read_csv("../venue.csv")

home_countries  = home_df[["country"]].copy()
away_countries  = away_df[["country"]].copy()
venue_countries = venue_df[["country"]].copy()
```

```python
all_countries = pd.concat(
    [home_countries, away_countries, venue_countries],
    ignore_index=True)
```

```python
before = len(all_countries)
all_countries = all_countries.dropna(subset=["country"])
print(f"[Info] Dropped {before - len(all_countries):,} null country rows")
```

```python
all_countries["country"] = (
    all_countries["country"]
    .str.strip()
    .str.title())

distinct_countries = all_countries["country"].drop_duplicates().reset_index(drop=True)
```

```python
print(f"\nDistinct countries in the dataset: {len(distinct_countries)}")
print("\nFull list:")
print(distinct_countries.sort_values().to_string())
```
