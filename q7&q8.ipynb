```python
import pandas as pd

df = pd.read_csv("statistics.csv")


aces_df = df[df["statistic_name"].str.lower() == "aces"].copy()

aces_df["home_value"] = pd.to_numeric(aces_df["home_value"], errors="coerce")
aces_df["away_value"] = pd.to_numeric(aces_df["away_value"], errors="coerce")

match_aces = aces_df.groupby("match_id")[["home_value", "away_value"]].sum()
match_aces["total_aces"] = match_aces["home_value"] + match_aces["away_value"]

avg_aces_per_match = match_aces["total_aces"].mean()
print(f"Average number of aces per match: {avg_aces_per_match:.2f}")
```

```python
import pandas as pd

stats_df = pd.read_csv("statistics.csv")
home_df = pd.read_csv("home_team.csv")
away_df = pd.read_csv("away_team.csv")

df_double_faults = stats_df[stats_df["statistic_name"].str.lower() == "double_faults"].copy()
df_double_faults["home_value"] = pd.to_numeric(df_double_faults["home_value"], errors="coerce")
df_double_faults["away_value"] = pd.to_numeric(df_double_faults["away_value"], errors="coerce")

match_df_totals = df_double_faults.groupby("match_id")[["home_value", "away_value"]].sum().reset_index()

home_gender = home_df[["match_id", "gender"]].rename(columns={"gender": "home_gender"})
away_gender = away_df[["match_id", "gender"]].rename(columns={"gender": "away_gender"})

match_df_totals = match_df_totals.merge(home_gender, on="match_id", how="left")
match_df_totals = match_df_totals.merge(away_gender, on="match_id", how="left")


same_gender_matches = match_df_totals[match_df_totals["home_gender"] == match_df_totals["away_gender"]].copy()
same_gender_matches["match_gender"] = same_gender_matches["home_gender"]

same_gender_matches["total_double_faults"] = (
    same_gender_matches["home_value"] + same_gender_matches["away_value"]
)


result = same_gender_matches.groupby("match_gender")["total_double_faults"].agg(
    ["mean", "median", "count"]
)

print("Average double faults per match by gender:")
print(result)
```
