```python
import pandas as pd
import zipfile
import os
import io
import re
from collections import defaultdict

# پوشه‌ای که همه فایل‌های zip توشه
base_path = r"C:\Users\Personal Camputer\OneDrive\Desktop\Tennis Schema"

# پوشه‌ای که فایل‌های csv خروجی ذخیره میشن (اگه نبود خودش میسازه)
output_path = r"C:\Users\Personal Camputer\OneDrive\Desktop\tennis"
os.makedirs(output_path, exist_ok=True)

# لیست فایل‌های zip
zip_files = [f for f in os.listdir(base_path) if f.endswith(".zip")]
print("تعداد فایل zip پیدا شده:", len(zip_files))
```

```python
data_by_category = defaultdict(list)

for zip_name in sorted(zip_files):
    zip_path = os.path.join(base_path, zip_name)
    with zipfile.ZipFile(zip_path, 'r') as z:
        for n in z.namelist():
            if not n.endswith(".parquet"):
                continue
            filename = os.path.basename(n)
            # اسم فایل رو از شماره آخرش جدا می‌کنیم تا اسم "دسته" به دست بیاد
            # مثلا home_team_11998445.parquet -> home_team
            category = re.sub(r'_\d+\.parquet$', '', filename)
            with z.open(n) as f:
                df = pd.read_parquet(io.BytesIO(f.read()))
                data_by_category[category].append(df)
    print(f"{zip_name} پردازش شد")
```

```python
print("\nدسته‌های پیدا شده:", list(data_by_category.keys()))
```

```python
for category, dfs in data_by_category.items():
    combined = pd.concat(dfs, ignore_index=True)
    out_file = os.path.join(output_path, f"{category}.csv")
    combined.to_csv(out_file, index=False)
    print(f"{category}: {len(combined)} ردیف -> ذخیره شد در {out_file}")
```
