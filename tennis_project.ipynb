**Question1**

  How many tennis players are included in the dataset?

```python
import pandas as pd

home_df = pd.read_csv('home_team.csv')
away_df = pd.read_csv('away_team.csv')

all_players = pd.concat([home_df, away_df])

unique_players = all_players.drop_duplicates(subset='player_id')
num_players = unique_players['player_id'].nunique()

print(f"Number of unique players: {num_players}")


```

**Question2**

What is the average height of the players?

```python
players_with_height = unique_players.dropna(subset=['height'])

average_height =round( players_with_height['height'].mean(),2)
print(f"players average height: {average_height}")
```

**Question3**

Which player has the highest number of wins?

```python
event_df = pd.read_csv('event.csv')

home_df_clean = home_df.drop_duplicates().drop_duplicates(subset='match_id', keep='first')
away_df_clean = away_df.drop_duplicates().drop_duplicates(subset='match_id', keep='first')
event_df_clean = event_df.drop_duplicates().drop_duplicates(subset='match_id', keep='first')

home_players = home_df_clean[['match_id', 'player_id', 'name']]
away_players = away_df_clean[['match_id', 'player_id', 'name']]

merged = event_df_clean.merge(home_players, on='match_id', suffixes=('', '_home'))
merged = merged.merge(away_players, on='match_id', suffixes=('_home', '_away'))

def get_winner(row):
    if row['winner_code'] == 1:
        return row['player_id_home']
    elif row['winner_code'] == 2:
        return row['player_id_away']
    else:
        return None

merged['winner_id'] = merged.apply(get_winner, axis=1)

win_counts = merged['winner_id'].value_counts()

top_player_id = win_counts.idxmax()
top_player_wins = win_counts.max()

all_players_clean = pd.concat([home_df_clean, away_df_clean]).drop_duplicates(subset='player_id')
top_player_name = all_players_clean[all_players_clean['player_id'] == top_player_id]['name'].values[0]

print("\nPlayer with most wins:", top_player_name)
print("Player ID:", top_player_id)
print("Number of wins:", top_player_wins)
```
